### Spieler
Klasse mit allen wichtigen Variablen
- name und sname (String) sind Vorname (name) und Nachname (surname)
- wpn (String) ist der Waffenname
- mag (String) ist der Name der Magie
- hp (int) sind die Lebenspunkte (Health Points) des Spielers, mit maxhp als Maximalwert. Wenn (hp <= 0)  --> tot
- sp (int) sind die Ausdauerpunkte (Ehrlich gesagt keine Ahnung warum sp aber ich werde das jetzt nicht mehr alles umschreiben), mit maxsp als Maximalwert. Wenn (sp <= 0)  --> Keine Aktion möglich (Physikal macht keinen Schaden)
- ma (int) sind die Manapunkte des Spielers. Wenn (ma <= 0)  --> Keine Aktion möglich (Magisch macht keinen Schaden)
- gld (int) ist die Goldtasche des Spielers, der ist wie viel Gold der Spieler hat
- magdam (int) ist, wie viel magischen Schaden der Spieler anrichtet
- wpndam (int) ist, wie viel physischen Schaden der Spieler anrichtet

```python
class Player:
    def __init__(self, name, sname, wpn, mag, hp, maxhp, sp, maxsp, ma, maxma, gld, magdam, wpndam):
        self.name = name
        self.sname = sname
        self.wpn = wpn
        self.mag = mag
        self.hp = hp
        self.maxhp = maxhp
        self.sp = sp
        self.maxsp = maxsp
        self.ma = ma
        self.maxma = maxma
        self.gld = gld
        self.magdam = magdam
        self.wpndam = wpndam
```

### Sich ausruhen
(Nur einmalig pro Zug und Maximalwerte können nicht überschritten werden)
- Regeneriert 5sp
- Regeneriert 1ma

```python
    def rest(self):
        self.sp += 5
        self.ma += 1
        self.statchk()
        print(f"{self.name} hat sich ausgeruht")
        time.sleep(0.2)
        console.print("+ 5sp", style = "green")
        time.sleep(0.2)
        console.print("+ 1ma", style = "green")
```

Die dazugehörige Funktion, um zu überprüfen ob ein Maximalwert überstiegen wurde:

```python
 def statchk(self):
        if self.hp > self.maxhp:
            self.hp = self.maxhp
        if self.sp > self.maxsp:
            self.sp = self.maxsp
        if self.ma > self.maxma:
            self.ma = self.maxma
        return
```

### Gegner
- enname (String) ist der Name des Gegners
- enhp (int) sind die Lebenspunkte (Healthpoints) des Gegners
- endam (int) ist der Schaden des Gegners pro Zug

```python
class Enemy:
    def __init__(self, enname, enhp, endam, engld):
        self.enname = enname
        self.enhp = enhp
        self.endam = endam
        self.engld = engld
```

### (Vorläufig) Gegner vom Prolog generieren lassen
Generiert zufällig einen Banditen oder ein Perlschwein. Es ist eine Chance von 2 / 3, gegen das schwächere Perlschwein zu kämpfen, und 1 / 3, gegen den stärkeren Banditen zu kämpfen

```python
def get_enemy1():
        en = random.randint(1, 3)
        if en == 1:
            enemy = Enemy(enname="Bandit", enhp=20, endam=random.randint(5, 8), engld=random.randint(5, 10))
            return enemy
        elif en == 2 or 3:
            enemy = Enemy(enname="Perlschwein", enhp=10, endam=random.randint(7, 8), engld=random.randint(1, 2))
            return enemy
```

### Kampfsystem (Wird noch ausgebaut)
#### Noch zu beheben:
   - Ausdauer kann unendlich verbraucht werden
   - Keine Möglichkeit, sich zu heilen
   - Mana kann unendlich verbraucht werden
   - Gegenstände (Wenn Inventar hinzugefügt)

Spieler und Gegner wählen abwechselnd eine Aktion für ihren Zug aus.

#### Der Spieler kann entscheiden zwischen:
  - Physischer Angriff (wpndam)
  - Magischer Angriff (madam)

Wenn hp <= 0, stirbt die respektive Partei.
Gewinnt der Spieler, bekommt der Spieler das Gold des Gegngers. Bei Tod des Spielers schließt sich das Spiel.

```python
def fighting1(pl):
    enemy = get_enemy1()
    finished = 0
    while finished == 0:
        if enemy.enhp <= 0:
            time.sleep(0.2)
            console.print("Besiegt!", style="yellow")
            pl.gld += enemy.engld
            time.sleep(0.2)
            console.print(f"Du hast {enemy.engld}gld gefunden! Du hast nun {pl.gld}gld")
            print("")
            print("~~~~~~~~~~~~~~")
            return pl
        print(f"Ein {enemy.enname} attackiert dich!")
        time.sleep(0.2)
        pl.hp -= enemy.endam
        console.print(f"{enemy.endam} Schaden genommen", style="red")
        time.sleep(0.2)
        print("")
        pl.stats()
        reacted = 0
        if pl.hp <= 0:
            console.print("Gestorben", style="red")
            input()
            return exit()
        enemy.enstats()
        while reacted == 0:
            CHOICE = f"""
                            Wie wehrst du dich?
                                - {pl.wpn} (Phy) - 5sp
                                - {pl.mag} (Mag) - 5ma
                            """
            chc = Markdown(CHOICE)
            print("~~~~~~~~~~~~~~")
            print(chc)
            print("~~~~~~~~~~~~~~")
            answer = input()
            print("")
            if answer.lower() == "phy":
                enemy.enhp -= pl.wpndam
                pl.sp -= 5
                console.print(f"Du hast {pl.wpndam} Schaden gemacht!", style="green")
                print("")
                reacted = 1
            elif answer.lower() == "mag":
                enemy.enhp -= pl.magdam
                pl.ma -= 5
                console.print(f"Du hast {pl.magdam} Schaden gemacht!", style="green")
                print("")
                reacted = 1
            else:
                print("Tipp: Siehe Inhalt Klammern")
```

### Befehlabfrage für Progress
Mögliche Optionen:
   - Weitergehen
        - Storyprogression
   - Ausruhen
      - Regeneriert +5sp und +1ma
   - Stats
      - Gibt die Stats des Spielers im Terminal aus (Siehe Methode weiter oben)
   - Suizid
      - Simples exit()

```python
def action1(pl):
        rested = 0
        actch = Markdown("""
        Was möchstest du tun?
            - Weiter (Weitergehen)
            - Ausruhen (+5sp, +1ma)
            - Stats (Zeigt Stats)
            - Suizid
        """)
        act = 0
        while act == 0:
            console.print(actch)
            print("~~~~~~~~~~~~~~~~~~~~~~~~~~~~")
            choi = input()
            print("")
            if choi.lower() == "weiter":
                act = 1
                fighting1(pl)
            elif choi.lower() == "stats":
                pl.stats()
            elif choi.lower() == "ausruhen" and rested == 0:
                pl.rest()
                rested = 1
            elif choi.lower() == "ausruhen" and rested == 1:
                console.print("Es ist zu gefährlich, sich so lange auszuruhen", style = "red")
            elif choi.lower() == "suizid":
                exit()
            else:
                console.print("Wie bitte?", style="red")
                print("")
                print("~~~~~~~~~~~~~~~~~~~~~~~~~~~~")
```
